# My first call

In this first lab we are testing that we have set up our environment properly. We will use openai's library to make a call to the LLM provider of our choice.

> [!INFO]
> Remember that openai api library defaults to openai's endpoints, to switch to a different provider we need to provide a respective url when we create the client.

```python
%load_ext autoreload
%autoreload 2
```

```python
import os

from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
from providers import (
    ANTHROPIC_API_MODEL,
    ANTHROPIC_API_URL,
    OPENROUTER_API_MODEL,
    OPENROUTER_API_URL,
)
```

```python
load_dotenv()
```

```python
question = "I am feeling a curious today. Tell me a historical fact about a region in Africa that is not widely known. Keep it short. Preferrably within a short paragraph. Use markdown format for headers"
```

## Antropic Endpoint

```python
anthropic_api_key = os.environ.get("ANTHROPIC_API_KEY", "")
print(f"ANTHROPIC_API_KEY = {anthropic_api_key[:5]}...")
```

```python
anthropic_client = OpenAI(api_key=anthropic_api_key, base_url=ANTHROPIC_API_URL, )
```

```python
anthropic_response = anthropic_client.chat.completions.create(
    model=ANTHROPIC_API_MODEL,
    messages=[
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user", "content": question}
    ],
)
```

```python
display(Markdown(anthropic_response.choices[0].message.content))
```
